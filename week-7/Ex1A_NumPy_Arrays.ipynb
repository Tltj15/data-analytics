## Looping through arrays

```python
#Create a variable named arrA as a two-dimensional array
#containing (11, 12, 13) and (14, 15, 16) then print arrA.
arrA = np.array([(11, 12, 13), (14, 15, 16)])
print(arrA)
# used the comma to separte the group of list numbers
```

```python
#Loop through the first dimension of arrA (i.e. for x in arrA) and
#print each element multiplied by 2.

for x in arrA:
    print(x * 2)
# x is the number in the group list * 2
```

```python
# loop through the second dimension of arrA (i.e. for y in x) and
#print each element multiplied by 2.
for x in arrA:
    for y in x:
        print(y * 2)

```

```python
# the first loop it loops through by each row [], []
# and in the second loop for y, its going throuhg each number individualy as a row
#basically printed in a vertical line vs horizontal group
```

```python
#Again loop through the second dimension of arrA (i.e. for y
#in x), but this time, assign x*2 to a variable named loop_result. Add another line
#of code to print loop_result, then add a comment to the cell to explain how/why
#this result is different.
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# output was repeated three times per group of numbers
```

```python
print(loop_result)
# it only gave an output of the second set of numbers [14,15,16]
#its outside of the loop 
```
