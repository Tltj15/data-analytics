## Creating basic plots

Create two variables containing the following data points:
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

Use plt.plot() to plot the line graph for x and y, and then display it using plt.show().

```python
plt.plot(x, y)
plt.show()
#plt.show is like print
```

Duplicate the previous code cell, then update it to add markers to the data points
using the marker='o' parameter

```python
plt.plot(x, y, marker='o')
#i perfer the 'D', mfc='r'(color), ms=10(markersize)
plt.show()
#'o' just shows markers no line
```

Duplicate the previous code cell, then customize the plot with a title (‘Example
Graph’), x-axis label (‘X axis’), and y-axis label (‘Y axis’)

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x, y, marker='o')
plt.show()
```

## Styling Plots

Plot the same data as above, but this time, change the line style to dashed ('--') and
the line color to red.

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x, y, marker='o', ls='--')
plt.show()
```

Change the marker style to a square ('s'), set the marker size to 10, and change the
marker color to green.

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x, y, marker='s', ms=10, mfc='g', ls='--')
plt.show()
```

## Advanced plotting techniques

Create a bar chart showing the sales of different products. Use the following data:
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
plt.bar(products, sales)
plt.show()
```

Customize your bar chart by adding a title, x-axis label, and y-axis label. Set the
color of the bars to 'skyblue' and adjust the bar width to 0.5.

```python
plt.title('Sales of Products')
plt.xlabel('Product')
plt.ylabel('Sales')

plt.bar(products, sales, color='skyblue', width=0.5 )
plt.show()
```

Create a scatter plot to visualize the relationship between two variables. Use the
following data:
x = np.random.rand(50)
y = np.random.rand(50)

```python
import numpy as np
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y)
plt.show()
```

Customize your scatter plot by setting the size of the markers based on another
variable (e.g., s = np.random.rand(50) * 100), and using a colormap (e.g.,
cmap='viridis')

```python

size = np.random.rand(50) *100
plt.title('Scatter Plot Example')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.scatter(x, y, s=size, c=size, cmap='viridis')
plt.colorbar() # add the color label
plt.show()
```

## Subplots

Create a figure with two subplots: one for a line chart and one for a histogram

```python
fig, (ax1, ax2) = plt.subplots(2)
```

In the first subplot, plot a sine wave using:
x = np.linspace(0, 10, 100)
y = np.sin(x)

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

In the second subplot, create a histogram of 1000 random numbers generated
using np.random.randn(1000)

```python
data = np.random.randn(1000)
```

Add titles and labels to each subplot

```python
ax1.plot(x, y)
ax1.set_title('Sine wave')
ax1.set_xlabel('X axis')
ax1.set_ylabel('sin(x)') # have to usee sin(x)for y axis

ax2.hist(data, bins=30) # divides data to 30 bins
ax2.set_title('Histogram')
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')


plt.tight_layout()# stops plots from overlaping
plt.show()
```

```python
# in order to show graph i have plt.show all together
fig, (ax1, ax2) = plt.subplots(2)
x = np.linspace(0, 10, 100)
y = np.sin(x)
data = np.random.randn(1000)
ax1.plot(x, y)
ax1.set_title('Sine wave')
ax1.set_xlabel('X axis')
ax1.set_ylabel('sin(x)') # have to usee sin(x)for y axis
ax2.hist(data, bins=30) # divides data to 30 bins
ax2.set_title('Histogram')
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')

plt.tight_layout()# stops plots from overlaping
plt.show()
```
