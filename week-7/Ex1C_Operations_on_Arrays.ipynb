## Rounding

```python
#Print r_3 to display the values you’ll be working with for the next few cells.
r_3 = np.random.rand(2, 1)
print(r_3)
```

```python
#Round the values in r_3 using trunc
result = np.trunc(r_3)
print(result)
#returns number with decimal place chopped off
```

```python
#Round the values in r_3 using fix
result = np.fix(r_3)
print(result)
#rounds to integer close to zero
```

```python
#Round the values in r_3 using floor
result = np.floor(r_3)
print(result)
```

```python
#Round the values in r_3 using ceil.
result = np.ceil(r_3)
print(result)
#rounds up to integer
```

```python
#Round the values in r_3 using round, to 2 decimal places.
result = np.round(r_3, 2) #2 means 2 decimal places
print(result)
# if no (r_3,2) it will round to whole numbers[0.][1.]
```

## Cumulative operations

```python
#Print r_4 to display the values you’ll be working with for the next few cells.
r_4 = np.random.randint(20, 100, size=(4,5))
print(r_4)
```

```python
#Print the sum of all values in r_4
result = np.sum(r_4) #total of all elements
print(result)
```

```python
#Print the sum of elements along the 0 axis for r_4.
result = np.sum(r_4, axis=0)
print(result) # adds down the coulmns each separatly output 5, 5 coloumns
```

```python
#Print the sum of elements along the 1 axis for r_4
result = np.sum(r_4, axis=1)
print(result) # out put 4, 4 rows
```

```python
#Print the cumulative sum (rolling total) of values in r_4.
result = np.cumsum(r_4)
print(result)
#
```

```python
#Print the cumulative sum (rolling total) of values along the 0 axis for r_4
result = np.cumsum(r_4, axis=0)
print(result)
```

Create two new variables, bag1 and bag2. The first variable represents the
contents of a grocery bag, which contains two vegetables, three fruits, four boxes of
snacks, and no beverages. The second variable represents another grocery bag,
which contains no fruits, four vegetables, two beverages, and one box of snacks.
Assign an appropriate array to each variable that represents how many of each
category of item are in each bag. Print bag1 and bag2

```python
bag1 = [2, 3, 4, 0]
# veggies, fruit, snack, bev
bag2 = [4, 0, 1, 2]
print(bag1)
print(bag2)
```

```python
#Write an expression to calculate the total number of items in each bag, and print the
#result in the form of a sentence: The first bag contains __ items, and the second bag
#contains ___ items.
total_items = np.sum(bag1)
total_items2 = np.sum(bag2)
print(f'The first bag contains {total_items} items, and the second bag contains {total_items2} items')


```

```python
#Write an expression to calculate the total number of items in each category, and
#print the result in the form of a sentence: There are __ vegetables, __ fruits, __ boxes
#of snacks, and __ beverages in the two grocery bags.
veggie = bag1[0] + bag2[0]
fruits = bag1[1] + bag2[1]
snacks = bag1[2] + bag2[2]
bev = bag1[3] + bag2[3]

print(f'There are {veggie} vegetables, {fruits} fruits, {snacks} boxes of snacks, and {bev} beverages in the two grocery bags')
```

## Teamwork

The scenario for this set of problems is that you are volunteering with a voter registration campaign, and you have been asked to oversee a squad of door-knockers. There are three people on your team: Ali, Ben, and Carla. Your squad went out to knock on doors each evening last week, Monday through Friday, and now you are assembling that data.

```python
#Ali reported knocking on 5 doors on Monday, 4 on Tuesday, 6 on Wednesday, 4 on
#Thursday, and 2 on Friday. Create a variable holding an array of these values.

ali = np.array([5, 4, 6, 4, 2])
print(ali)
```

```python
#Ben missed Monday because he was out of town. He reported knocking on 5 on
#Tuesday, 5 on Wednesday, 3 on Thursday, and 4 on Friday. Create a variable holding
#an array of Ben’s values.
ben = np.array([0, 5, 5, 3, 4])
print(ben)
```

```python
#Carla reported knocking on 3 doors on Monday, 6 on Tuesday, 3 on Thursday, and 4
#on Friday. She was sick on Wednesday so missed that day. Create a variable
#holding an array of Carla’s values.
carla = np.array([3, 6, 0, 3, 4])
print(carla)
```

```python
#Compile the values reported by Ali, Ben, and Carla into a single array containing all
#door-knocks for the week, as all_knocks. Print the resulting array.
all_knocks = np.array([ali, ben, carla])
print(all_knocks)

```

```python
#Use all_knocks to calculate the total number of doors that each person knocked
#on for the week, and also the total number of doors that the group collectively
#knocked on. Print the results.
np.sum(all_knocks, axis=1)
np.sum(all_knocks)
total_num = np.sum(all_knocks, axis=1) # axis 1 adds accross row vs axis 0 adds down the columns
group_total = np.sum(all_knocks)
print(total_num, 'each person on team')
print(f'Ali knocked {total_num[0]}') #[0],[1],[2] index value of each person
print(f'Ben knocked {total_num[1]}')
print(f'Carla knocked {total_num[2]}')
print(group_total, 'group total knocks')

```

The organization you’re working with estimates that for every two door-knocks, one of those people will actually turn out to vote. For your report to the organization, what is the rolling total of estimated votes that your group was able to turn out, day by day for last week? (Hint: Start with the rolling total by day, then multiply that array by an appropriate value to get the result.)

```python
all_knocks = np.array([ali, ben, carla])

day_by_day = np.sum(all_knocks, axis=0) #adds the rows for each day total
rolling_total = np.cumsum(day_by_day)
turn_out = rolling_total *2 #turn over every 2knocks one will vote
print(turn_out)
```
