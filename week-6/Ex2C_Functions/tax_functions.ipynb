# Calculating taxes 
# S.S tax, Medicare tax, Federal tax

```python
def get_soc_sec_tax(gross_pay):
    tax_rate = .062 #6.2%
    money = gross_pay
    amount = money * tax_rate
    print(f'The social security tax amount is $ {amount:.2f}')

get_soc_sec_tax(gross_pay= 2000)

```

```python
def get_medicare_tax(gross_pay):
    med_tax_rate = .0145 #1.45%
    money = gross_pay
    amount = money * med_tax_rate
    print(f'The Medicare tax amount is $ {amount:.2F}')

get_medicare_tax(gross_pay= 15000)
```

```python
#Withholding Code Tax Rate
#0 23%
#1 21%
#2 19.5%
#3 18.5%
#4+ 18% less 0.5% for each withholding over 4 (ex, code 6: rate .18 – (2 x .005) = .17 (or 17%)
def get_federal_tax(gross_pay, tax_code): # using the sample withholding code and tax rate
    money = gross_pay
    if tax_code == 0:
        tax = .23 
    elif tax_code == 1:
        tax = .21
    elif tax_code == 2:
        tax = .195
    elif tax_code == 3:
        tax = .185
    elif tax_code == 4:
        tax = .18
    else:
        tax = .18 - (2 * .005) #any number above 4 will get this tax 

    amount = money * tax
    print(f'The Federal tax amount is $ {amount:.2f}')

get_federal_tax(gross_pay= 5000, tax_code= 6)        


```
