This notebook analyzes a small sales dataset using Python functions. Each function takes the full dataset as it's first argument.

```python
sales_data = [
('Laptop', 'East', 1200.00),
('Phone', 'West', 850.00),
('Tablet', 'East', 400.00),
('Monitor', 'South', 650.00),
('Keyboard', 'West', 75.00),
('Laptop', 'South', 1150.00),
('Phone', 'East', 900.00),
('Tablet', 'West', 425.00),
('Monitor', 'East', 680.00),
('Keyboard', 'South', 80.00),
('Laptop', 'West', 1300.00),
('Phone', 'West', 820.00),
]
```

## Lab 1. Total sales by region

```python
def region_total(data, region):
    return sum(amt for _, r, amt in data if r == region) # sum amount for sales in region
#the _ is a convention for a variable, but do not need to use
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    return sum(1 for _, r, _ in data if r == region) # replace amt with 1 and the the second amt with _
```

```python
print(region_count(sales_data, 'East'))
print(region_count(sales_data, 'West'))
print(region_count(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):
    total = region_total(data, region)
    count = region_count(data, region)
    return total / count if count > 0 else 0 # had issues have to have everything on one line
```

```python
print(region_average(sales_data, 'East'))
print(region_average(sales_data, 'West'))
print(region_average(sales_data, 'South'))
print(region_average(sales_data, 'North'))
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold): #return a new list, only the records
    # where sale amount > threshold
    return [record for record in data if record[2] > threshold] #2 is the index for sale amount
```

```python
threshold = 500
result = sales_above(sales_data, threshold)
for record in result: 
    print(record)
```

## Lab 5. Top and bottom sale

```python
def top_sale(data): # tuple with the highest sale amount
    return max(data, key=lambda s: s[2]) #index 2 is the sale amount 

def bottom_sale(data):
    return min(data, key=lambda s: s[2])    
```

```python
print(top_sale(sales_data))
print(bottom_sale(sales_data))
```

The top sale was a Laptop in the West region for $1300.0

## Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):
    return list(map(lambda s: (s[0], s[1], s[2] * (1 - discount_pct)), data)) # uses index 0,1,2 to only take the first three in the list
#using the list from sales data 
# with map to apply the lambda to every item in the list
```

```python
discount_data = apply_discount(sales_data, .15)
# fucntion for new data set with .15 discount

```

```python

print(sales_data[:3]) # sales data list
print(discount_data[:3]) # new list with the discount amount
```

## Lab 7. Unique regions

```python
def unique_regions(data):
    return sorted(set(s[1] for s in data)) #1 is the postion for region in the list(0product,2price)
    # sorted organize 
#set removes dulpicates
```

```python
print(unique_regions(sales_data))
```

The function went through the data list for region, then removed duplicates and sorted them alphabetically 

## Lab 8. Region summary

```python
def region_summary(data, region):
    #use list comprehension to call region
    region_data = [s for s in data if s[1] == region]
    total = region_total(data, region) 
    count = region_count(data, region)
    avg = region_average(data, region)
    top = top_sale(region_data) # used the region_data for top sale
    # if i used top_sale(data) it gave me the top sale disregarding the region

    print(f'==={region} Region ===')
    print(f'Records: {count}')
    print(f'Total sales: $ {total:.2f}')
    print(f'Average sale: $ {avg:.2f}')
    print(f'Total sale: {top[0]} - ${top[2]:.2f}')

region_summary(sales_data, 'East')
```

## Lab 9. Full Summary

```python
regions = unique_regions(sales_data)#use unique function
for region in regions:
    region_summary(sales_data, region)

# using the unique fucntion that sorted and removed duplicates im able to get the three regions
#using the region summary to get the print format for all three regions
#used a for loop
```
