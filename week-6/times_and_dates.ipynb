```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
#current date+time
print(today)
```

```python
#use the .strftime()
day_of_week = today.strftime('%a')
month = today.strftime('%b')
day_month = today.strftime('%d')
year = today.strftime('%y')
#%a returns day of the week short version
#%b month name short version
#%d day of month
#%y year short version

print('Today is' + ' ' + day_of_week + ',' + ' ' + month  + ' '+ day_month + ' ' + ',' + year)
```

# My birthday

```python
## use datetime.date() for brithday
birth_day = datetime.date(1997, 9, 15)
print(birth_day)
```

```python
#use .strftime() MM/DD/YYYY
month = birth_day.strftime('%m')
#%m month as a number
day = birth_day.strftime('%d')
#%d day of month
year = birth_day.strftime('%Y')
#%Y upper Y full year

print('My birthday is' + ' ' +  month + '/' + day + '/' + year)
```

# Next 90 Days

```python
#use datetime.timedelta
ninety_d = today + datetime.timedelta(days=90)
```

```python
# print month day, year
month = ninety_d.strftime('%b')
#%b month short version
day = ninety_d.strftime('%d')
#%d day of month
year = ninety_d.strftime('%y')

print(f'90 days from today is {month} {day}, {year}')
```

# Time for Dinner

```python
dinner_time = datetime.time(18,0)
print(dinner_time)
```

```python
# print HH:MM PM
hour = dinner_time.strftime('%I')
#%I hour in civilian time
min = dinner_time.strftime('%M')
#%M minutes 
p_m = dinner_time.strftime('%p')
#%p means am/ pm 


print(f'Lets meet for dinner at {hour}:{min} {p_m}')
# found out i can code like this as well
#print(f'Lets meet for dinner at {dinner_time.strftime('%I:%M %p')}')
```
