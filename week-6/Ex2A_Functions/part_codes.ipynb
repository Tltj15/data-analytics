## Lab 2. parsing string - part codes

```python
ACME:123-L supplied by Acme, part #123, size large
DI:12345-M supplied by DI, part #12345, size medium
ACE:1-12 supplied by Ace, part #1, size 12
```

```python
#return all characters before the :
def get_supplier_code(part_code):
    for i in range(len(part_code)):
        if part_code[i] == ':':
            supplier = part_code[:i]
            return supplier


print(get_supplier_code('ACME:123-L'))
print(get_supplier_code('DI:12345-M'))
print(get_supplier_code('ACE:1-12'))

```

```python
# return all characters between the : and -
def get_product_num(part_code):
    start = 0 # for :
    end = 0 #for -
    for i in range(len(part_code)):
        if part_code[i] == ':':
            start = i +1
        if part_code[i] == '-':
           end = i
    product = part_code[start:end]
    return product
# had to create variables bc the last print only had one number 
#with the two if statement that targets the last print and the first two work with the first if
print(get_product_num('ACME:123-L'))
print(get_product_num('DI:12345-M'))
print(get_product_num('ACE:1-12'))
```

```python
#return all characters after the -
def get_size(part_code):
    for i in range(len(part_code)):
        if part_code[i] == '-':
            size = part_code[i+1:]
    return size
# i had issues with figuring out where to put the :
# at first i had size=part_code[i+1] and that only gave me 1 for the last print
#the correct answer had to be 12
print(get_size('ACME:123-L'))
print(get_size('DI:12345-M'))
print(get_size('ACE:1-12'))
```

```python
print('For ACME:123-L the supplier is', get_supplier_code('ACME:123-L'), 'part number', get_product_num('ACME:123-L'), 'is available in size', get_size('ACME:123-L'))
```
